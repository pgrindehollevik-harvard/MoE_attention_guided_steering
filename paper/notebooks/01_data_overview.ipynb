# 01 — Data and prompt-contract overview

Sanity check the fear-concept dataset and the two evaluation contracts
(prefix-conditioned and question-only) used by the paper. Reads
`experiments/<run_dir>/manual_review_plan.json` directly; no model weights.

```python
import json
from pathlib import Path

REPO = Path('.').resolve()
while REPO != REPO.parent and not (REPO / 'experiments').exists():
    REPO = REPO.parent

main_run = REPO / 'experiments' / 'mixtral_steermoe_fears_seed7_question_only_with_llama'
plan = json.loads((main_run / 'manual_review_plan.json').read_text())
print('concept family:', plan.get('concept_family'))
print('evaluation family:', plan.get('evaluation_family'))
print('seed:', plan.get('seed'))
print('concepts:', plan.get('sampled_concepts'))
print('eval prompts:')
for q in plan.get('sampled_evaluation_questions', []):
    print(' ', q)
```

## Prompt contracts

```python
cases = plan.get('cases', [])
print('total cases:', len(cases))
if cases:
    case = cases[0]
    print('--- Case 0 ---')
    print('concept:', case.get('concept'))
    print('eval question:', case.get('evaluation_question'))
    print('prefix-conditioned diagnostic prompt:', case.get('prompt_text'))
    print('question-only test prompt:', case.get('full_prompt_text', case.get('evaluation_question')))
```
