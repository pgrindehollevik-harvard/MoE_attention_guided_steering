# 02 — Prefix-conditioned model comparison (Stage 1)

Loads the committed full-prefix comparison report and renders one
(concept, eval) row at a time. This is the qualitative evidence behind
Claim 1: Mixtral 8x7B follows explicit fear prefixes.

```python
import json
from pathlib import Path

REPO = Path('.').resolve()
while REPO != REPO.parent and not (REPO / 'experiments').exists():
    REPO = REPO.parent

candidates = [
    'prefix_conditioned_model_comparison_fears_seed7_mixtral_8bit',
    'prefix_conditioned_model_comparison_fears_seed7',
]
run_dir = next(
    (REPO / 'experiments' / c for c in candidates if (REPO / 'experiments' / c).exists()),
    None,
)
results = json.loads((run_dir / 'model_comparison_results.json').read_text())
print('run dir:', run_dir.name)
print('total cases:', len(results.get('cases', [])))
```

```python
from textwrap import shorten
for case in results.get('cases', [])[:5]:
    print(f"\n[{case['concept']}] eval v{case['evaluation_version']}")
    print('prompt:', shorten(case['prompt_text'], 140))
    for tag, text in case.get('responses', {}).items():
        print(f'  {tag:>32}: {shorten(text, 200) if text else "(empty)"}')
```
