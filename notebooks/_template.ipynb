## Provenance

Write `experiments/<run_dir>/provenance.json` so the run is
self-describing. Subprocess to `git` is the simplest portable way.

```python
import json
import socket
import subprocess
import time

out_dir = Path(output_dir)
out_dir.mkdir(parents=True, exist_ok=True)
git_sha = subprocess.run(['git', 'rev-parse', 'HEAD'], capture_output=True, text=True).stdout.strip()
provenance = {
    'notebook': '<replace with notebook filename>',
    'run_name': run_name,
    'output_dir': str(output_dir),
    'seed': seed,
    'limit_cases': limit_cases,
    'model_id': model_id,
    'git_sha': git_sha,
    'host': socket.gethostname(),
    'slurm_job_id': os.environ.get('SLURM_JOB_ID'),
    'wallclock_utc': time.strftime('%Y-%m-%dT%H:%M:%SZ', time.gmtime()),
}
(out_dir / 'provenance.json').write_text(json.dumps(provenance, indent=2))
print('wrote', out_dir / 'provenance.json')
```
